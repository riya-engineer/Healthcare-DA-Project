```python
HealthDisease = df.groupby('SleepTime')['BMI'].sum()

plt.bar(
    HealthDisease.index,   # x-axis (categories)
    HealthDisease.values   # y-axis (heights)
)

plt.title('Health Disease')
plt.xlabel('SleepTime')
plt.ylabel('BMI Sum')
plt.show()
```

```python
import matplotlib.pyplot as plt

GenHealth = df.groupby('GenHealth')['Race'].count()  # better than sum for categories

plt.pie(
    GenHealth,                 # ✅ use the grouped data
    labels=GenHealth.index,    # ✅ labels from index
    autopct='%.1f%%',
    startangle=90,
    wedgeprops={'width': 0.6, 'edgecolor': 'black'}  # donut style
)

plt.title('GenHealth')
plt.axis('equal')
plt.show()
```
